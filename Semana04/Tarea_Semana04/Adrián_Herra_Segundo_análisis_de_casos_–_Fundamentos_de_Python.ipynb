# Problema #1: Calculadora de propina inteligente

## Problema

- Se necesita, crear un programa, que funcione como calculadora de propina inteligente, lo que significa que se requeire que se solicite al usuario el nombre del cliente, el total de la cuenta a pagar y el porcentaje de propina para el mesero, por último se mostrara en pantalla el sub-total a pagar, propina a pagar y total a pagar.

## Entrada
- nombre del cliente (texto)
- saldo cuenta (float)
- porcentaje propina (float)

## Proceso
- monto propina = monto cuenta *  (porcentaje propina / 100)
- saldo total = saldo cuenta + monto propina

## Salida esperada
- Mostrar nombre cliente saldo cuenta + monto propina = saldo total

```python
# Autor: Adrián Herra Vargas
# Programa Calculadora de Propina Inteligente

#Entradas
nombre_cliente = input("Ingrese el nombre del cliente: ").upper().strip()
try:
    saldo_cuenta = float(input("Ingrese el monto de la cuenta: "))
    if saldo_cuenta < 1:
        print(f"\tError el saldo es incorrecto {saldo_cuenta} ")
except ValueError:
    print(f"\nMonto ingresado invalido {saldo_cuenta}, favor ingrese un cantidad valida")
    saldo_cuenta = -1
try:
    porcentaje_propina = int(input("Ingrese el porcentaje de la propina: "))
    if porcentaje_propina < 1:
        print(f"\tPorcentaje de propina invalida {porcentaje_propina}, digite del 1 a 100")
except ValueError:
    print(f"\tMonto ingresado invalido {porcentaje_propina}, digite del 1 a 100")
    porcentaje_propina = -1
#Procesos

if not nombre_cliente or saldo_cuenta < 1 or porcentaje_propina < 1:
    resultado = "\tDatos erroneos, no es posible mostrar el resultado de la propina..!"
else:
    propina = saldo_cuenta * (porcentaje_propina / 100)
    saldo_total = saldo_cuenta + propina
    resultado = f"La propina del cliente {nombre_cliente} es ¢{propina} \nEl saldo total a pagar seria de: ¢{saldo_total}"

#Salida
print("-" * 50, '\n')
print(resultado)
```

# Problema #2:
Detector de número par/impar

## Problema

Dado el ingreso de un número entero, se debe identificar si el mismo es par o impar, validando que la entrada sea valida mediante el control del error númerico y aplicando un condicional que separa que tipo de número es.

## Entrada

- numero (int)

## Proceso

- Validar si la entrada es valida
- Si el residuo de numero es igual a 0 entonces
     respuesta = par
  sino respuesta = par

## Salida esperada

- Imprimir resultado

```python
# Autor: Adrián Herra Vargas
# Programa Detector de número par/impar

#Entradas
try:
    numero = int(input("Ingrese el número: "))
    if numero < 1:
        print(f"\tError el número ingresado no es valido {numero}, ingrese un número mayor que 0")
except ValueError:
    print(f"\nFormato de número invalido, favor ingrese un número entero valido")
    numero = -1

#Procesos

if numero < 1:
    resultado = "\tNo fue posible realizar la detección, intente de nuevo"
else:
    formato = numero % 2 == 0
    if formato:
        resultado = f"El número {numero} es par"
    else:
        resultado = f"El número {numero} es par"

#Salida
print("-" * 70, '\n')
print(resultado)

```

# Problema #3: Clasificador de edad

## Problema

Se requiere ingresar un número entero el cual se designara como la edad,
la cual será validad en menor de edad si el rango es de 1 a 18 años cumplidos,
adulto si la edad mayor o igual a 18 años y adulto mayor si es mayor a 65 años

## Entrada

- edad (int)

## Proceso

- Validar si edad tiene un formato valido, y tambien validar si edad = 18
entonces es menor de edad sino si es mayor o igual a 18 = adulto, sino es adulto mayor.

## Salida esperada

- Desplegar en pantalla el resultado de la clasificación de edad

```python
# Autor: Adrián Herra Vargas
# Programa Clasificador de edad

#Entradas
try:
    edad = int(input("Ingrese la edad: "))
    if edad < 1:
        print(f"\tError la edad ingresada no es valida {edad}, digite del 1 a 120")
except ValueError:
    print(f"\tError formato invalido, ingrese una edad valida")
    edad = -1

#Procesos

if edad < 1:
    resultado = "\tNo fue posible clasificar la edad..!"
else:
    if edad < 18:
        resultado = f"La edad ingresada {edad} es de una persona: Menor de edad"
    elif edad >= 65:
        resultado = f"La edad ingresada {edad} es de una persona: Adulto(a) mayor"
    else:
        resultado = f"La edad ingresada {edad} es de una persona: Adulto(a)"

#Salida
print("-" * 70, '\n')
print(resultado)

```

# Problema #4: Validador de compra con descuento

## Problema

Se requiere realizar un programa que valide el monto de las compras de los clientes, además si el cliente compra más de un millon adquiere descuento automatico o si es cliente frecuente y su compra es mayor a diez mil colones,
el resultado del programa mostrara

## Entrada

-nombre del cliente (texto)
-monto de la compra (float)
-tipo de cliente  (si/no) (texto)

## Proceso

- validar si la entrada es valida sino mostrar mensaje de error, ademas validar si es cliente frecuente y la compra >= 10000 entonces mostrar el nombre del cliente y el mensaje de que cuenta con un descuento aprobado, sino mostrar el nombre del cliente y que no cuenta con descuento la compra.

## Salida esperada

- motrar el resultado de las validadiones del proceso.

## Pruebas manuales

nombre = adrian
monto compra = 4500
cliente frecuente = si



```python
# Autor: Adrián Herra Vargas
# Programa Validador de compra con descuento

#Entradas
nombre_cliente = input("Ingrese el nombre del cliente: ").upper().strip()
try:
    monto = float(input("Ingrese el monto de la compra: "))
    if monto < 1:
        print(f"\tError el monto {monto} debe ser mayor a 1")
except ValueError:
    print(f"\tMonto ingresado invalido, favor ingrese un cantidad valida")
    monto = -1

cliente_frecuente = input("Es cliente frecuente (si/no)?: ").lower().strip() == 'si'

#Procesos

if not nombre_cliente or monto < 1:
    resultado = "Registro rechazado: revise los datos ingresados!"
elif cliente_frecuente and monto > 10000 or monto >= 1000000:
    resultado = f"Descuento aprobado para {nombre_cliente}: monto de compra ¢{monto}"
else:
    resultado = f"Estimado cliente {nombre_cliente}: monto de compra ¢{monto}, no aplica descuento,\nSolo aplica para compras mayores a ¢10000 y clientes frecuentes"


#Salida
print("-" * 80, '\n')
print(resultado)
```
